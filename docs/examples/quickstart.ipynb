# PyMyCorr Quick Start

This notebook demonstrates how to use the pymycorr client to fetch data from the MyCorr API.

## Setup

Set your API token as an environment variable:

```bash
export MYCORR_API_TOKEN="your-api-token"
```

Optionally, you can also override the default API URL:

```bash
export MYCORR_API_URL="https://custom-api.example.com"
```

Or create a `.env` file in your project root with these values.

````python
from pymycorr import MyCorr

# Initialize the client (automatically loads from .env file and environment variables)
client = MyCorr()

# Or initialize with explicit parameters:
# client = MyCorr(url="https://custom-api.example.com", token="your-token")

# Or specify a custom .env file:
# client = MyCorr(env_file="/path/to/.env")
````

## Fetch Data as pandas DataFrame

````python
# Replace with your actual table ID
TABLE_ID = "<your table id>"

# Fetch the latest version as a pandas DataFrame (default)
df = client.get_table(TABLE_ID)
display(df)

# You can also specify a version number or alias:
# df = client.get_table(TABLE_ID, version=2)        # specific version
# df = client.get_table(TABLE_ID, version="stable") # version alias
````

## Fetch Data as polars DataFrame

````python
# Fetch as polars DataFrame
df_polars = client.get_table(TABLE_ID, engine="polars")
display(df_polars)
````

## Get Table Info

````python
# Get table metadata
info = client.get_table_info(TABLE_ID)
info
````
