# PyMyCorr Quick Start

This notebook demonstrates how to use the pymycorr client to fetch data from the MyCorr API.

## Setup

First, set your API token as an environment variable:

```bash
export MYCORR_TOKEN="your-api-token"
export MYCORR_URL="https://api.mycorr.com/data/table"
```

Or create a `.env` file in your project root with these values.

````python
import os

from dotenv import load_dotenv

# Load environment variables from .env file
load_dotenv()

MYCORR_URL = os.getenv("MYCORR_URL")
MYCORR_TOKEN = os.getenv("MYCORR_TOKEN")

if not MYCORR_URL or not MYCORR_TOKEN:
    raise ValueError("Please set MYCORR_URL and MYCORR_TOKEN environment variables")
````

````python
from pymycorr import MyCorr

# Initialize the client
client = MyCorr(url=MYCORR_URL, token=MYCORR_TOKEN)
````

## Fetch Data as pandas DataFrame

````python
# Replace with your actual table ID
TABLE_ID = "your-table-id"

# Fetch the latest version as a pandas DataFrame
df = client.get_table(TABLE_ID, version="latest", engine="pandas")
display(df)
````

## Fetch Data as polars DataFrame

````python
# Fetch as polars DataFrame
df_polars = client.get_table(TABLE_ID, engine="polars")
display(df_polars)
````

## Get Table Info

````python
# Get table metadata
info = client.get_table_info(TABLE_ID)
print(f"Table ID: {info['table_id']}")
print(f"Rows: {info['num_rows']}")
print(f"Columns: {info['column_names']}")
print(f"Schema: {info['schema']}")
````
